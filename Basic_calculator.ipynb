```python
n1=int(input("enter num1:"))
n2=int(input("enter num2:"))
op=input("give operator:")
if op=='+':
  print(f"addition of 2 numbers is:{n1+n2}")
elif op=='-':
  print(f"subtraction of 2 numbers is:{n1-n2}")
elif op=='*':
  print(f"multiplication of 2 numbers is:{n1*n2}")
elif op=='/':
  print(f"division of 2 numbers is:{n1/n2}")
else:
  print("invalid operator")
```
